```python
# Basic python libraries - numeric calculations, dataframes, and plotting
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# The dataset for this activity
from sklearn.datasets import load_iris

# The ML libraries
from sklearn.linear_model import LogisticRegression
from sklearn.cluster import KMeans

# ML Methodology
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

Let's do Machine Learning on this data now

Difference between *Supervised*, *Unsupervised*, and *Self-supervised* learning

### Supervised, Unsupervised, and Self-Supervised Learning

**Supervised learning:** I would use the flower measurements as the inputs and the species names as the labels. I could train a model to predict the species of a new flower.

**Unsupervised learning:** I would ignore the species labels and use a method like K-Means to group flowers that have similar measurements. Then I could compare the groups with the real species afterward.

**Self-supervised learning:** I would make a learning task from the Iris data itself. For example, I could hide one measurement, such as petal width, and train a model to predict it from the other measurements. This lets the model learn patterns without using the species labels.


```python
# K-Means on Iris — no labels used during fitting
kmeans = KMeans(n_clusters=3, n_init=10)
cluster_labels = kmeans.fit_predict(X_iris)

# Visualize on petal length vs. petal width (the most separable pair)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Left: true species labels
for species, color, idx in zip(iris.target_names,
                                ['steelblue', 'tomato', 'seagreen'],
                                [0, 1, 2]):
    mask = y_iris == idx
    axes[0].scatter(X_iris[mask, 2], X_iris[mask, 3],
                    c=color, s=25, alpha=0.8, label=species)
axes[0].set_xlabel('petal length (cm)')
axes[0].set_ylabel('petal width (cm)')
axes[0].set_title('True Species Labels (supervised)')
axes[0].legend()

# Right: k-means clusters (no labels used)
cluster_colors = ['steelblue', 'tomato', 'seagreen']
for k in range(3):
    mask = cluster_labels == k
    axes[1].scatter(X_iris[mask, 2], X_iris[mask, 3],
                    c=cluster_colors[k], s=25, alpha=0.8, label=f'Cluster {k}')
axes[1].scatter(kmeans.cluster_centers_[:, 2], kmeans.cluster_centers_[:, 3],
                c='black', marker='X', s=150, zorder=5, label='Centroids')
axes[1].set_xlabel('petal length (cm)')
axes[1].set_ylabel('petal width (cm)')
axes[1].set_title('K-Means Discovered Clusters (unsupervised)')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# Try different numbers of clusters
for n_clusters in [2, 3, 4]:
    km = KMeans(n_clusters=n_clusters, n_init=10, random_state=42)
    labels = km.fit_predict(X_iris)

    plt.figure(figsize=(6, 4))
    for k in range(n_clusters):
        mask = labels == k
        plt.scatter(X_iris[mask, 2], X_iris[mask, 3], s=25, alpha=0.8, label=f"Cluster {k}")

    plt.scatter(km.cluster_centers_[:, 2], km.cluster_centers_[:, 3],
                c='black', marker='X', s=150, label='Centroids')
    plt.xlabel('petal length (cm)')
    plt.ylabel('petal width (cm)')
    plt.title(f'K-Means with {n_clusters} Clusters')
    plt.legend()
    plt.show()
```

### Changing the Number of Clusters

I tried K-Means with 2, 3, and 4 clusters. With **2 clusters**, some flowers that belong to different species get grouped together. With **3 clusters**, the groups match the Iris data better because the dataset has three species. With **4 clusters**, one of the natural groups gets split into smaller groups.

If I did not know how many Iris species there were, I could try several values of `k` and use a method such as the **elbow method** or **silhouette score** to help choose a good number of clusters.


Generalization

```python
# Train/test split for Iris — 80/20 split
X_train, X_test, y_train, y_test = train_test_split(
    X_iris, y_iris, test_size=0.2
)

# Fit logistic regression
clf = LogisticRegression(max_iter=200)
clf.fit(X_train, y_train)

iris_lr_ein  = 1 - accuracy_score(y_train, clf.predict(X_train))
iris_lr_eout = 1 - accuracy_score(y_test,  clf.predict(X_test))

print(f'Logistic Regression on Iris:')
print(f'  E_in  (120 training samples) = {iris_lr_ein:.3f}')
print(f'  E_out  (30 test samples)     = {iris_lr_eout:.3f}')
print(f'  Generalization gap            = {iris_lr_eout - iris_lr_ein:.3f}')
```

### Generalization Gap

The generalization gap is the difference between `E_out` and `E_in`. `E_in` is the training error, and `E_out` is the test error.

In my run, the gap is small, so the model performs similarly on the training data and on new test data. A **high generalization gap** would mean the model does much better on the training data than on new data. That usually means the model is **overfitting**.

To reduce a high gap, I could use more training data, use a simpler model, or use regularization.
